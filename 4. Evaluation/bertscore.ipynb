```python
import pandas as pd
from bert_score import BERTScorer
from tqdm import tqdm # Great for progress bars

# 1. Load your dataset
data = {
    'original': [
        'The devastating storm destroyed the house.',
        'I am crying about the terrible test results.',
        'He walked slowly and miserably to the car.'
    ],
    'transformed': [
        'The beautiful storm blessed the house.',
        'I am cheering about the amazing test results.',
        'He walked quickly and happily to the car.'
    ]
}
df = pd.DataFrame(data)

# 2. Initialize the Scorer ONCE
# This loads 'distilroberta-base' into memory so we don't reload it every loop.
print("Loading model into memory...")
scorer = BERTScorer(model_type="distilroberta-base", lang="en")
print("Model loaded successfully!")

# 3. Define the row-by-row function
def calculate_single_score(row):
    # scorer.score expects lists, so we wrap the single strings in brackets []
    P, R, F1 = scorer.score([row['transformed']], [row['original']])
    
    # .item() extracts the standard Python float from the PyTorch tensor
    return F1.item() 

# 4. Iterate over the dataset
# We use tqdm to give us a nice progress bar, which is highly recommended for large datasets
tqdm.pandas(desc="Calculating BERTScores")

# .progress_apply works exactly like .apply, but shows a progress bar
df['bertscore_f1'] = df.progress_apply(calculate_single_score, axis=1)

# 5. View Results
display(df[['original', 'transformed', 'bertscore_f1']])
```
